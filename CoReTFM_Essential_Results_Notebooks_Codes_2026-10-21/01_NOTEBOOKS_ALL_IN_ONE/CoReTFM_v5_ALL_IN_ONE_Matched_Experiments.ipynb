```python
# 7 — GPU and model-version preflight; secret never printed or archived
import importlib.util
GPU_INFO=q.cuda_status()
print('GPU:',GPU_INFO)
TOKEN_AVAILABLE=bool(os.environ.get('TABPFN_TOKEN','').strip())
if IN_COLAB and LOOKUP_TABPFN_COLAB_SECRET and not TOKEN_AVAILABLE:
    try:
        from google.colab import userdata
        token=userdata.get('TABPFN_TOKEN')
        if token and str(token).strip():
            os.environ['TABPFN_TOKEN']=str(token).strip()
            TOKEN_AVAILABLE=True
        del token
    except Exception as ex:
        print('Colab secret lookup unavailable:',type(ex).__name__,
              '— continue TabICL; use Secrets sidebar when ready.')
print('TabPFN access:', 'TOKEN_CONFIGURED (license still required)' if TOKEN_AVAILABLE else 'NOT_CONFIGURED')
print('TabPFN-3 license acknowledged in notebook:',TABPFN_LICENSE_ACCEPTED)

def ensure_gpu_model(model):
    if not q.cuda_status().get('cuda'):
        return {'status':'SKIPPED_NO_CUDA','model':model}
    if model not in MODEL_PINS:return {'status':'UNKNOWN_MODEL','model':model}
    package,wanted=MODEL_PINS[model]
    current=q.version(package)
    if current!=wanted:
        if not AUTO_INSTALL_GPU_MODELS:return {'status':'MODEL_VERSION_MISMATCH','found':current,'required':wanted}
        print('Installing pinned',package,'version',wanted)
        subprocess.run([sys.executable,'-m','pip','install','-q',f'{package}=={wanted}'],check=True)
        current=q.version(package)
    if current!=wanted:return {'status':'PIN_UNSATISFIED','found':current,'required':wanted}
    if model=='tabpfn3':
        from tabpfn.constants import ModelVersion
        if not hasattr(ModelVersion,'V3'):
            return {'status':'BLOCKED_MISSING_MODELVERSION_V3'}
    return {'status':'PACKAGE_READY','version':current}
```

### Interpreting the result correctly
